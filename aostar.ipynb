```python
graph={
    'A':[('B','C'), ('D',)],
    'B':['E',],
    'C':['F',],
    'D':['G'],
    'E':[],
    'F':[],
    'G':[],
}
h={
    'A':10,
    'B':4,
    'C':3,
    'D':6,
    'E':2,
    'F':1,
    'G':2,

}

cost={
    ('A','B'):1,
    ('A','C'):1,
    ('A','D'):2,
    ('B','E'):2,
    ('C','F'):2,
    ('D','G'):1
}
solution={}
def ao_star(node):
    if not graph[node]:
      return h[node]
    print("\nExpanding:",node)
    min_cost_for_node = float('inf')

    for option in graph[node]:
      current_option_cost = 0

      for child in option:
        child_cost=ao_star(child)
        current_option_cost += cost [(node,child)]+child_cost

      min_cost_for_node = min(min_cost_for_node, current_option_cost)

    return min_cost_for_node
```

```python
graph = {
    'A': [('B', 'C'), ('D',)],
    'B': [('E',)],
    'C': [('F',)],
    'D': [('G',)],
    'E': [],
    'F': [],
    'G': []
}

h = {
    'A': 10,
    'B': 8,
    'C': 5,
    'D': 7,
    'E': 3,
    'F': 6,
    'G': 4
}

cost = {
    ('A', 'B'): 1,
    ('A', 'C'): 1,
    ('A', 'D'): 2,
    ('B', 'E'): 2,
    ('C', 'F'): 2,
    ('D', 'G'): 1
}

solution = {}


def ao_star(node):
    # If node is a leaf node
    if not graph[node]:
        return h[node]

    print("\nExpanding:", node)

    best_cost = float('inf')
    best_option = None

    # Check all possible options
    for option in graph[node]:
        total = 0

        for child in option:
            child_cost = ao_star(child)
            total += cost[(node, child)] + child_cost

        print("Option", option, "Cost =", total)

        if total < best_cost:
            best_cost = total
            best_option = option

    # Update heuristic value
    h[node] = best_cost
    solution[node] = best_option

    print("Best choice for", node, "=", best_option)
    print("Updated cost of", node, "=", best_cost)

    return best_cost


# Run AO* algorithm
ao_star('A')

print("\n--------------------")
print("Optimal solution:")
for node, option in solution.items():
    print(node, "->", option)

print("\nUpdated heuristic costs:")
for node, value in h.items():
    print("h[" + node + "] =", value)
```
